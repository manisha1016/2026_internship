```python
import pandas as pd #import pandas
```

```python
df = pd.read_csv("messy_sales_data.csv") #read data
```

```python
print(df.head()) #First 5 row display
```

```python
print(df.shape)  #display total row and column
```

```python
df.describe()
```

```python
print(df.isna().sum()) #missing value identify
```

```python
print(df.duplicated().sum()) #duplicate value identify
```

```python
df = df.drop_duplicates() #remove duplicate
```

```python
df = df.dropna(subset=["customer_name", "order_date"]) #remove duplicate
```

```python
df["quantity"] = df["quantity"].fillna(df["quantity"].median())  #missing value insert
```

```python
df["unit_price"] = df["unit_price"].fillna(df["unit_price"].mean())
```

```python
df["rating"] = df["rating"].fillna(df["rating"].mean())
```

```python
df["payment_method"] = df["payment_method"].fillna("Unknown")
```

```python
df["city"] = df["city"].fillna("Unknown")
```

```python
df["email"] = df["email"].fillna("not_provided")
```

```python
print(df.isna().sum()) #return missing value identify
```

```python
df["order_date"] = pd.to_datetime(df["order_date"], format="mixed", dayfirst=True)  #date and time format change
```

```python
df["city"] = df["city"].str.strip().str.title()  #uppercase,lowercase ahange
```

```python
df.to_csv("cleaned_sales_data.csv", index=False) #save file
```
