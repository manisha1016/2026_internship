```python
import pandas as pd
df = pd.read_csv("cleaned_orders.csv")
```

```python
df["payment_method"] = df["payment_method"].str.replace("cod", "cash on delivery", case=False)
print(df["payment_method"].unique())
```

```python
# Convert all category names into uppercase

df["category"] = df["category"].str.upper()

# Display unique categories
print(df["category"].unique())
```

```python
# Convert payment methods into uppercase

df["payment_method"] = df["payment_method"].str.upper()

# Check updated values
print(df["payment_method"].unique())
```
