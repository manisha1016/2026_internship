```python
df = df.drop_duplicates()  #remove duplicate
```

```python
print(df.duplicated().sum())
```

```python
df["customer_name"] = df["customer_name"].str.strip() #extra space remove
```

```python
df["state"] = df["state"].str.strip()
```

```python
df["category"] = df["category"].str.strip()
```

```python
df["product"] = df["product"].str.strip()
```

```python
df["payment_method"] = df["payment_method"].str.strip()
```

```python
df["order_status"] = df["order_status"].str.strip()
```

```python
df["gender"] = df["gender"].replace({ "M": "Male","F": "Female"}) #replace value
```

```python
df["email"] = df["email"].str.strip().str.lower() #email lowercase
```

```python
df["payment_method"] = df["payment_method"].str.strip().str.lower() #payment method lowercase
```

```python
df["order_status"] = df["order_status"].str.strip().str.lower()
```

```python
print(df.head())
```

```python
print(df.isnull().sum()) #find duplicate values
```

```python
df["age"] = df["age"].astype("Int64") #age flote to int
```

```python
df["rating"] = df["rating"].astype("Int64") #rating float to interger
```

```python
df["unit_price"] = df["unit_price"].astype(str).str.replace(r"[₹,Rs.\s]", "", regex=True) #Unit Price Number Format
```

```python
df["unit_price"] = pd.to_numeric(df["unit_price"], errors="coerce")
```

```python
df.isnull().sum()
```

```python
print(df.duplicated().sum())
```

```python
df = df.dropna()  # Remove missing values
```
